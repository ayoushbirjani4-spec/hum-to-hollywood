## Option B: the full web app with a public link

```
import os, importlib.util
os.chdir('/content/hth')
os.environ['HTH_BACKEND'] = 'musicgen'
# Load app/app.py by path: a plain `import app` finds the app/ folder instead of the file
spec = importlib.util.spec_from_file_location('hth_app', '/content/hth/app/app.py')
hth_app = importlib.util.module_from_spec(spec)
spec.loader.exec_module(hth_app)
hth_app.demo.launch(share=True, debug=True)
```
